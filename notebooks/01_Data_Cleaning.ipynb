```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../dataset/raw/House Price India.csv")

print("Dataset Loaded Successfully!")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# Check duplicate records
duplicates = df.duplicated().sum()

print(f"Total Duplicate Records: {duplicates}")
```

```python
df.dtypes
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("/", "_")
)

df.columns
```

```python
df.describe()
```

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

plt.figure(figsize=(18,8))

df[numeric_columns].boxplot(rot=90)

plt.title("Box Plot of Numerical Features")

plt.show()
```

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[(df["price"] >= lower_limit) &
        (df["price"] <= upper_limit)]

print("Dataset Shape After Removing Outliers:")
print(df.shape)
```

```python
CURRENT_YEAR = 2026

df["property_age"] = CURRENT_YEAR - df["built_year"]

df[["built_year", "property_age"]].head()
```

```python
print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
df.to_csv("../dataset/cleaned/clean_house_price_india.csv", index=False)

print("Clean dataset saved successfully!")
```
