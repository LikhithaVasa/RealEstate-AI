```python
plt.figure(figsize=(10,6))

plt.hist(df["price"], bins=40)

plt.title("House Price Distribution")

plt.xlabel("Price")

plt.ylabel("Count")

plt.show()
```

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(10,6))

plt.hist(df["living_area"], bins=30)

plt.title("Living Area Distribution")

plt.xlabel("Living Area")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(df["living_area"], df["price"])

plt.xlabel("Living Area")

plt.ylabel("Price")

plt.title("Living Area vs Price")

plt.show()
```

```python
correlation = df.corr(numeric_only=True)

correlation
```

```python
plt.figure(figsize=(15,10))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    annot=False
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
correlation["price"].sort_values(ascending=False)
```

```python
plt.figure(figsize=(12,6))

plt.hist(df["price"], bins=40, edgecolor="black")

plt.title("Property Price Distribution")

plt.xlabel("Price")

plt.ylabel("Number of Properties")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.hist(df["living_area"], bins=30, edgecolor="black")

plt.title("Living Area Distribution")

plt.xlabel("Living Area")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df["number_of_bedrooms"].value_counts().sort_index().plot(kind="bar")

plt.title("Number of Bedrooms")

plt.xlabel("Bedrooms")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df["number_of_bathrooms"].value_counts().sort_index().plot(kind="bar")

plt.title("Number of Bathrooms")

plt.xlabel("Bathrooms")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(df["living_area"], df["price"], alpha=0.5)

plt.title("Living Area vs Price")

plt.xlabel("Living Area")

plt.ylabel("Price")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x="number_of_bedrooms",
    y="price",
    data=df
)

plt.title("Bedrooms vs Price")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x="number_of_bathrooms",
    y="price",
    data=df
)

plt.title("Bathrooms vs Price")

plt.show()
```

```python
corr = df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(16,12))

sns.heatmap(
    corr,
    cmap="coolwarm",
    annot=False
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
corr["price"].sort_values(ascending=False)
```
